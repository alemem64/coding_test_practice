```python
from collections import deque

graph = {
    1: [2, 3],
    2: [4],
    3: [],
    4: []
}

def bfs(start):

    # 1. 출발점 저장
    q = deque([start])

    # 2. 출발점 방문 표시
    visited = {start}

    # 3. 탐색할 정점이 남아 있는 동안
    while q:

        # 4. 가장 먼저 발견한 정점 꺼내기
        node = q.popleft()

        print(node)

        # 5. 현재 정점과 연결된 정점 확인
        for next_node in graph[node]:

            # 6. 아직 방문하지 않았다면
            if next_node not in visited:

                # 7. 방문 표시
                visited.add(next_node)

                # 8. 나중에 탐색하도록 저장
                q.append(next_node)

bfs(1)
```

```python
from collections import deque
def bfs(si,sj):
    if arr[si][sj]>0:       # 현재위치가 최단거리
        return (si,sj)

    q = deque()
    v = [[0]*(N+1) for _ in range(N+1)]

    q.append((si,sj))
    v[si][sj]=1
    tlst = []

    while q:
        nq = deque()

        for ci,cj in q:
            for di,dj in ((-1,0),(1,0),(0,-1),(0,1)):
                ni,nj = ci+di, cj+dj
                # 네방향, 범위내, 미방문, *조건(불가능:-1,sset(청소기들 좌표) 가능: 0,>0)
                if 1<=ni<=N and 1<=nj<=N and v[ni][nj]==0 and arr[ni][nj]>=0 and (ni,nj) not in sset:
                    if arr[ni][nj]==0:          # 빈칸: 계속 진행
                        nq.append((ni,nj))
                        v[ni][nj]=1
                    else:                       # 먼지: >0
                        tlst.append((ni,nj))    # 같은 거리에 먼지들 더 있을 수 있으니

        if tlst:        # 오염된 격자(들) 찾은 경우
            tlst.sort() # 여러 개일 경우 행-열 오름차순
            return tlst[0]
        q = nq

    # 결국 오염된 격자를 못찾은 경우(제자리..)
    return (si,sj)

# 가능한 최대 청소량: 20
def clean(ci,cj):
    #   왼, 위, 오, 아 (청소를 skip할 위치 1개 지정)
    di=[ 0,-1, 0, 1]
    dj=[-1, 0, 1, 0]

    # [1] 청소를 skip할 방향 선정
    mx, skip = -1, -1
    for skip_dr in range(4):
        sm = 0
        for k in range(4):
            if k==skip_dr:  continue        #  skip
            ni,nj = ci+di[k], cj+dj[k]
            if 1<=ni<=N and 1<=nj<=N and arr[ni][nj]>0:
                sm += min(20, arr[ni][nj])  # 최대 20

        if mx<sm:
            mx=sm
            skip=skip_dr

    # [2] 청소진행: 최대 20만큼만 청소가능
    arr[ci][cj] -= min(20, arr[ci][cj])     # 청소기위치
    for k in range(4):                      # 네방향
        if k == skip:   continue  # skip
        ni, nj = ci + di[k], cj + dj[k]
        if 1 <= ni <= N and 1 <= nj <= N and arr[ni][nj] > 0:
            arr[ni][nj]-= min(20, arr[ni][nj])

def print_arr(arr):
    for lst in arr[1:]:
        print(*lst[1:], sep='\t')

def print_unit(unit):
    for i in range(1, K+1):
        print(unit[i], end=' ')

##############################################

N, K, L = map(int, input().split())
arr = [[0] * (N + 1)] + [[0] + list(map(int, input().split())) for _ in range(N)]

unit = {}                   # 번호별 좌표저장
sset = set()                # 청소기 좌표저장(중복확인)
ans = [0] * L               # 정답 저장
for i in range(1, K + 1):
    si, sj = map(int, input().split())
    unit[i] = [si, sj]
    sset.add((si, sj))

for l in range(L):          # L번 테스트 진행
    # [1] 번호순 가장 가까운 오염된 격자 이동(행,열순)
    # 현재위치 먼지 있으면: 제자리, 오염격자 없으면? 제자리

    # print(f'########## {l} turn ##########')
    # print_arr(arr)
    # print_unit(unit)

    for idx in range(1, K+1):
        ci,cj = unit[idx]
        ni,nj = bfs(ci,cj)
        if (ci,cj)!=(ni,nj):    # 청소기 이동
            unit[idx]=[ni,nj]
            sset.remove((ci,cj))
            sset.add((ni,nj))

    # [2] 번호순 4방향중 청소량이 가장 많은 구역 청소 진행
    # 청소할 수 있는 먼지량은 최대 20!!!!
    for idx in range(1, K+1):
        ci,cj = unit[idx]
        clean(ci,cj)            # 네 방향중 최대 청소가능 방향 청소!

    # [3] 먼지 축적: 먼지있는(>0) 모든 격자 +5
    for i in range(1,N+1):
        for j in range(1,N+1):
            if arr[i][j]>0:
                arr[i][j]+=5

    # [4] 먼지 확산(동시): 깨끗한 격자 주변 네 방향합 //10
    narr = [lst[:] for lst in arr]
    for i in range(1,N+1):
        for j in range(1,N+1):
            if arr[i][j]==0:
                sm = 0
                for di,dj in ((-1,0),(1,0),(0,-1),(0,1)):
                    ni,nj = i+di, j+dj
                    if 1<=ni<=N and 1<=nj<=N and arr[ni][nj]>0:
                        sm += arr[ni][nj]

                narr[i][j]+=sm//10
    arr=narr

    # [5] 먼지 합계 출력
    for i in range(1,N+1):
        for j in range(1,N+1):
            if arr[i][j]>0:
                ans[l]+=arr[i][j]

    if ans[l]==0:           # 모든위치 먼지 0
        break
print(*ans, sep='\n')
```
