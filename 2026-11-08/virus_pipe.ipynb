```python
from collections import deque

def solution(n, infection, edges, k):
    answer = 0
    
    # 감염된 노드 set
    sset = set(infection)
    best = 1
    
    # 노드별 연결된 노드 graph
    graph = [[] for _ in range(n+1)]
    for x,y,t in edges:
        graph[x].append((y,t))
        graph[y].append((x,t))
    
    def spread(sset,pipe):
        nset = set(sset)
        q = deque(sset)
        
        while q:
            cur = q.popleft()
            for nxt,t in graph[cur]:
                if nxt not in nset and t == pipe:
                    nset.add(nxt)
                    q.append(nxt)
                    
        return nset
        
    
    
    def dfs(sset,n_open):
        
        best = max(best,len(sset))
        
        # k번 열면 종료
        if n_open == k: return
        
        for pipe in range(1,4):
            # 감염시키기
            nset = spread(sset,pipe)
            
            # dfs 호출
            dfs(nset,n_open+1)
            
        return best
        
    
    answer = dfs(sset,0)
    
    return answer
```
