```python
def solution(dist_limit, split_limit):
    best = 1

    def dfs(nodes, used, last):              # last: 직전 층에서 쓴 수
        nonlocal best
        best = max(best, nodes)

        remain = dist_limit - used
        for c in (2, 3):
            if c < last:                     # ★ 3 다음에 2는 안 씀 (추가된 줄)
                continue
            if nodes * c > split_limit:
                continue
            if nodes <= remain:
                dfs(nodes * c, used + nodes, c)
            else:
                best = max(best, (nodes - remain) + remain * c)

    dfs(1, 0, 2)                             # 처음엔 2, 3 둘 다 가능
    return best
```
