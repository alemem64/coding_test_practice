````python
N, M = map(int, input().split())

# boxes[idx][num,h,w,c,r]
tbox = [ list(map(int,input().split())) for _ in range(M)]
boxes = {}
for idx in range(M):
    boxes[idx] = {"num": tbox[idx][0], "h": tbox[idx][1], "w": tbox[idx][2], "c": tbox[idx][3], "r":-N}
arr = [ [0]*(N+1) for _ in range(N+1) ]

ans = []

# [1] 택배 투입: i번째 순서대로 투입, 중력에 의해 아래쪽 택배부터 순서대로 떨어짐, 공간 벗어나지 않음 보장
# 바닥에 닿거나 다른 박스 만나면 멈춤
for idx in range(M): # M개 박스에 대해서
    landed = False
    for r in range(1, N+1): # 첫줄부터 N줄 중
        for w in range(boxes[idx]["w"]): # 박스 width만큼의 하단에
            if r+boxes[idx]["h"] > N or arr[r + boxes[idx]["h"]][boxes[idx]["c"] + w] > 0  :         # 아래에 택배가 존재하고 바닥을 벗어나면 해당 row로 멈춤
                
                # arr에 num을 반영
                for th in range(boxes[idx]["h"]):
                    for tw in range(boxes[idx]["w"]):
                        arr[r+th][boxes[idx]["c"]+tw] = boxes[idx]["num"] 
                    
                # boxes r 저장
                boxes[idx]["r"] = r
                
                # for 문 탈출
                landed = True
                break
        if landed: break

# print(boxes)
# print("처음 투입 \n", arr)

# [2] 택배 하차: 왼 오 반복, 빼는 길에 택배 없는 것 먼저, 그중 num 작은 거 먼저, 이후 떨어질 거 떨어짐
for order in range(M): # M개 박스에 대해 택배 하차
    
    can_out = set() # 나올 수 있는 박스 num들
    for idx in boxes:
        can_out.add(boxes[idx]["num"])
        
    # 부딪히지 않고 뺄 수 있는 택배 찾기
    for idx in boxes: 
        stuck = False
        for h in range(boxes[idx]["h"]): # 높이마다 왼쪽 끝이 비어 있는지 확인
            
            if order % 2 == 0: # 좌측으로 뺄 때
                for c in range(1, boxes[idx]["c"]): # 왼쪽 1칸씩 비교
                    if arr[boxes[idx]["r"] + h][ boxes[idx]["c"] - c ] > 0: # 다른 택배가 있다면
                        can_out.remove(boxes[idx]["num"])
                        stuck = True
                        break
                if stuck: break

            else: # 우측으로 뺄 때
                for c in range(1, N + 2 - boxes[idx]["w"] - boxes[idx]["c"]): # 오른쪽 1칸씩 비교
                    if arr[boxes[idx]["r"] + h][ boxes[idx]["c"] + boxes[idx]["w"] + c -1 ] > 0: # 다른 택배가 있다면
                        can_out.remove(boxes[idx]["num"])
                        stuck = True
                        break
                if stuck: break

    # 그중 num이 작은 거 하차
    out = min(can_out)
    ans.append(out)
    for i in boxes:
        if boxes[i]["num"] == out:
            del boxes[i]
            break
    for i in range(N+1):
        for j in range(N+1):
            if arr[i][j] == out:
                arr[i][j] = 0
        
    # print("택배를 뺌 \n", arr)
    
    # 떨어질 수 있는 택배 떨어짐
    # boxes를 하단 row가 큰 순서대로 (아래 있는 순서로 정렬)
    sorted_idx = sorted(boxes, key= lambda idx: boxes[idx]["r"] + boxes[idx]["h"] - 1, reverse=True)
    
    for idx in sorted_idx:
        landed = False
        for r in range(boxes[idx]["r"], N+2 - boxes[idx]["h"]): # 바닥에 있는 각 줄마다
            for w in range(boxes[idx]["w"]): # 박스 width만큼의 하단에
                if r+boxes[idx]["h"] > N or arr[r + boxes[idx]["h"]][boxes[idx]["c"] + w] > 0:         # 아래에 택배가 존재하거나 바닥을 벗어나면 해당 row로 멈춤
                    # print("order", order, "번호", boxes[idx]["num"], "r", r)
                          
                    # 기존 위치를 arr에서 제거
                    for i in range(N+1):
                        for j in range(N+1):
                            if arr[i][j] == boxes[idx]["num"]:
                                arr[i][j] = 0
                    
                    # 새 위치를 arr에 저장
                    for th in range(boxes[idx]["h"]):
                        for tw in range(boxes[idx]["w"]):
                            arr[r+th][boxes[idx]["c"]+tw] = boxes[idx]["num"] 
                    
                    # boxes r 저장
                    boxes[idx]["r"] = r
                
                    # for 문 탈출
                    landed = True
                    break
            if landed: break

    # print("택배가 떨어짐 order:", order, "\n", arr)

print(*ans, sep="\n")
````

````python
def find_bottom(si,sj,ei,ej):
    for i in range(ei,N+1):
        if sum(arr[i][sj:ej])!=0:       # 박스가 있다면
            return i
    return N+1

def mark(num,si,sj,ei,ej):
    for i in range(si,ei):
        arr[i][sj:ej]=[num]*(ej-sj)

def drop(num):
    si,sj,ei,ej = unit[num]

    bottom_i = find_bottom(si,sj,ei,ej)
    n_si,n_ei = bottom_i-(ei-si),bottom_i

    mark(0,si,sj,ei,ej)                 # 번호 지우기
    mark(num,n_si,sj,n_ei,ej)           # 번호 표시

    unit[num]=[n_si,sj,n_ei,ej]         # unit 갱신


# 위로 올라가면서 만나는 박스 중력처리
def gravity(si,sj,ei,ej):
    sset = set()
    for i in range(si-1, 0, -1):
        for j in range(1, N+1):
            if arr[i][j]!=0 and arr[i][j] not in sset:    # 중복 안 된 박스면
                drop(arr[i][j]) # 그 박스 중력처리
                sset.add(arr[i][j])

##################################################
N, M = map(int, input().split())

unit = {}
v = [0]*101         # 존재하는 박스
arr = [[0]*(N+1)]+[[0]*(N+1) for _ in range(N)]
# [1] 택배 투입(직사각형, 중력, 공간 벗어나지 않음)
for _ in range(M):
    # 번호,높이,너비,열번호
    k, h, w, c = map(int, input().split())

    si,sj,ei,ej = 1,c,1+h,c+w

    bottom_i = find_bottom(si,sj,ei,ej)     # 바닥좌표 찾기
    n_si,n_ei = bottom_i-h,bottom_i
    mark(k,n_si,sj,n_ei,ej)                 # 번호 표시

    unit[k]=[n_si,sj,n_ei,ej]
    v[k]=1

# [2] 택배하차(좌우 반복처리, 번호순, 뺀 후 중력처리)
# M개박스: 오름차순으로 존재하는 박스중 왼쪽/오른쪽 삭제 후 중력
ans = []
left = True
for _ in range(M):
    for num in range(1, 101):
        if v[num]==0:       continue    # 없는 박스면 skip

        si,sj,ei,ej=unit[num]
        for i in range(si, ei):
            if left==True:              # 왼쪽처리
                if sum(arr[i][1:sj])!=0:    # 장애물 있음: 택배 하차 불가능
                    break
            else:                       # 오른쪽하차
                if sum(arr[i][ej:N+1])!=0:  # 장애물 있음: 택배 하차 불가능
                    break
        else:                           # 모두 0: 하차가능
            # 기존택배 제거
            mark(0,si,sj,ei,ej)
            v[num]=0
            ans.append(num)

            # 중력처리
            gravity(si,sj,ei,ej)
            break
    left=not left                       # 왼-오-왼-오..

print(*ans, sep='\n')
````
