# 배운 점
1. k를 늘려가며 세지 않아도 됨. 작은 k에서 패턴 나오면 큰 k에서 이미 포함됨.
2. int(문자열, 2)로 2진수 변환 가능
3. python에서 문자열보다 숫자가 메모리 적게 차지
4. dic.get(key, default)로 key 없으면 default 반환, key 있으면 value 반환

```python
S = input()
K, M = map(int, input().split())

n = len(S)

# Please write your code here.

unstable = False

# 1. 패턴 길이를 K부터 1씩 늘리며 n까지 확인 -> k
for k in range(K,min(60,n)):

    # 2. 패턴 시작점을 첫 칸부터 n - k까지(포함) 잡고 해당 패턴으로 비교를 고정
    for si in range(n-k+1):
        p = S[si:si+k]
        cnt = 1

        # 3. 시작점 다음 칸부터 n - k(포함)까지 해당 패턴 길이와 비교해서 개수 세기
        for bi in range(si+1,n-k+1):
            if p == S[bi:bi+k]:
                cnt +=1
        
        # 4. 개수가 M개 이상이면 1 출력, 아니면 2로 돌아가기
        if cnt >= M:
            print(1)
            unstable = True
            break
    
    if unstable:
        break

if not unstable: print(0)

```
