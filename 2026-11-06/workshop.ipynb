```python
from collections import deque

def bfs(si,sj,arr):
    h = [[0]*N for _ in range(N)]
    h[si][sj] = 1

    r = [[0]*N for _ in range(N)]
    r[si][sj] = (si,sj)

    k = [[0]*N for _ in range(N)]

    
    hmax = []
    q = deque()
    q.append((si,sj))
    
    # K칸 이동하면서 h에 해당 칸의 인접 높이, r에 이전 칸 기록.
    # K칸 도달시 좌표 돌아가면서 인접 높이 큰 것을 hmax 리스트에 갱신
    # 모든 hmax 리스트에서 최소값 return
    while q:
        ci,cj = q.popleft()

        # k칸 도달시
        if k[ci][cj] == K-1:
            
            tmax = []
            tmax.append(h[ci][cj])

            # 해당 경로의 인접 높이 최대값
            while (ci,cj) != (si,sj):
                ci,cj = r[ci][cj]
                tmax.append(h[ci][cj])
            
            hmax.append(max(tmax))


        for di,dj in ((-1,0),(1,0),(0,-1),(0,1)):
            ni,nj = ci+di, cj+dj

            # 격자내, 미방문(h가 0이면), 조건(다음 칸이 더 높음)
            if 0<=ni<=N-1 and 0<=nj<=N-1 and h[ni][nj] == 0 and arr[ni][nj] > arr[ci][cj]:
                h[ni][nj] = arr[ni][nj] - arr[ci][cj]
                r[ni][nj] = (ci,cj)
                k[ni][nj] = k[ci][cj] + 1
                q.append((ni,nj))

    # print(k)
    if hmax: return min(hmax)
    else: return float('inf')

####################

N, K = map(int, input().split())
arr = [list(map(int, input().split())) for _ in range(N)]

# Please write your code here.
sarr = [[float('inf')]*N for _ in range(N)]

# 1. NxN 칸 중 시작 위치 선택
for si in range(N):
    for sj in range(N):

        # 2. 해당 위치에서 K칸만큼 이동
        # 이동시 높이가 높아져야함
        # 높이 차들간 최댓값 중 최소값 저장

        # bfs(si,sj,arr)
        sarr[si][sj] = bfs(si,sj,arr)


# 3. 각 시작 칸 중 가장 최댓값 작은 것 출력

ans = float('inf')
for i in range(N):
    for j in range(N):
        if ans > sarr[i][j]:
            ans = sarr[i][j]

if ans == float('inf'): print(-1)
else: print(ans)
```

```python
import sys
sys.setrecursionlimit(10000)

def dfs(ci, cj, length, cur_max):
    global ans
    # 가지치기: 이미 찾은 답보다 크거나 같으면 더 볼 필요 없음
    if cur_max >= ans:
        return
    # 길이 K 도달 → 답 갱신 (더 가도 최댓값은 줄지 않으므로 종료)
    if length >= K:
        ans = cur_max
        return

    for di, dj in ((-1,0),(1,0),(0,-1),(0,1)):
        ni, nj = ci+di, cj+dj
        # 격자 내, 다음 칸이 더 높음 (높이가 계속 증가하므로 방문 배열 불필요)
        if 0<=ni<N and 0<=nj<N and arr[ni][nj] > arr[ci][cj]:
            dfs(ni, nj, length+1, max(cur_max, arr[ni][nj]-arr[ci][cj]))

####################

N, K = map(int, input().split())
arr = [list(map(int, input().split())) for _ in range(N)]

ans = float('inf')

# 모든 칸을 시작점으로 시도
for si in range(N):
    for sj in range(N):
        dfs(si, sj, 1, 0)

if ans == float('inf'): print(-1)
else: print(ans)
```

```python
def next_layer(h, arr):
    # 다음 길이의 h를 새로 만듦 (아직 도달 못한 칸은 inf)
    nh = [[float('inf')]*N for _ in range(N)]

    for ci in range(N):
        for cj in range(N):

            # 현재 길이로 (ci,cj)에 도달 불가능하면 건너뜀
            if h[ci][cj] == float('inf'):
                continue

            for di,dj in ((-1,0),(1,0),(0,-1),(0,1)):
                ni,nj = ci+di, cj+dj

                # 격자내, 조건(다음 칸이 더 높음)
                # 높이가 계속 높아지므로 방문 체크 불필요
                if 0<=ni<=N-1 and 0<=nj<=N-1 and arr[ni][nj] > arr[ci][cj]:

                    # 지금까지 경로의 최댓값 vs 이번에 이동할 때의 높이 차
                    hmax = max(h[ci][cj], arr[ni][nj] - arr[ci][cj])

                    # 더 작은 최댓값으로 도착하는 경우만 갱신
                    if nh[ni][nj] > hmax:
                        nh[ni][nj] = hmax

    return nh

####################

N, K = map(int, input().split())
arr = [list(map(int, input().split())) for _ in range(N)]

# Please write your code here.

# h[i][j]: 길이 k인 등산로가 (i,j)에서 끝날 때,
#          인접 높이 차 최댓값 중 최소값

# 1. 길이 1: 어느 칸에서 시작해도 상관 없음, 높이 차 없음 → 0
h = [[0]*N for _ in range(N)]

# 2. 길이 2, 3, ..., K까지 한 칸씩 이동하며 h 갱신
for k in range(2, K+1):
    h = next_layer(h, arr)

# 3. 길이 K로 끝나는 칸들 중 최댓값 작은 것 출력
# (K보다 긴 등산로도 앞의 K칸만 보면 최댓값이 같거나 작으므로 K만 보면 됨)
ans = float('inf')
for i in range(N):
    for j in range(N):
        if ans > h[i][j]:
            ans = h[i][j]

if ans == float('inf'): print(-1)
else: print(ans)
```
