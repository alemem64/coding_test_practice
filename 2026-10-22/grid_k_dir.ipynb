이 격자판의 (1,1)칸에서 출발하여 도착 지점인 (N,N)칸에 도달하려 합니다.

초기 상태에서는 오른쪽 또는 아래쪽으로 방향을 정해 출발하며, 출발 이후에는 한 번에 한 칸씩 현재 바라보고 있는 방향으로 이동합니다.

각 이동 시, 현재 바라보고 있는 방향으로 계속 직진하거나 방향을 전환할 수 있습니다.

방향을 전환할 경우, 현재 오른쪽을 바라보고 있다면 아래쪽을 바라보도록, 현재 아래쪽을 바라보고 있다면 오른쪽을 바라보도록 전환됩니다. 위를 보거나 왼쪽을 보는 것은 불가능하다는 것을 주의해야 합니다.

최대 K번의 방향 전환을 사용하여 시작 지점에서 도착 지점까지 도달하는 방법의 수를 출력하는 프로그램을 작성하세요.

이 네 가지 방법 외에 최대 두 번의 방향 전환으로 도달하는 방법은 없으므로, 정답은 4입니다.
