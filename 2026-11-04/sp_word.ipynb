문제 설명
카카오톡은 메시지의 일부를 가려두었다가, 클릭했을 때만 공개되는 스포 방지 기능을 제공합니다. 이 기능을 활용하면 중요한 정보를 가리고 보낼 수 있습니다.

무지는 이 기능을 이용해 하나의 메시지 곳곳에 스포 방지 기능을 적용해 당신에게 보냈습니다. 당신은 메시지 시작부터 왼쪽 → 오른쪽 순서로 스포 방지 구간을 하나씩 클릭해 공개되는 단어들 중, 중요한 단어가 몇 개인지 확인하려 합니다.

단어 및 중요한 단어 규칙

단어는 공백으로 구분되며, 알파벳 소문자와 숫자로만 구성된 연속된 문자열입니다.
단어를 구성하는 문자들의 인덱스 중 하나 이상이 스포 방지 구간에 포함될 경우, 해당 단어는 스포일러 방지 단어로 간주합니다. 즉, 단어 내 일부 문자에만 스포일러 방지 기능이 적용되더라도, 해당 단어 전체를 스포일러 방지 단어로 간주합니다.
한 단어가 여러 개의 스포 방지 구간에 걸쳐 있을 수 있으며, 하나의 스포 방지 구간에 여러 단어가 포함될 수 있습니다.
스포 방지 구간을 클릭해 단어의 모든 문자가 공개되었을 때, 그 단어가 아래 조건을 모두 만족하면 중요한 단어입니다.
스포 방지 단어여야 합니다.
메시지의 스포 방지 구간이 아닌 구간(= 어떤 스포 방지 구간에도 속하지 않는 모든 구간: 각 구간의 앞·사이·뒤 포함)에 등장한 적이 없어야 합니다.
이전에 공개된 스포 방지 단어와 중복되지 않아야 합니다.
여러 단어가 동시에 공개된 경우, 왼쪽부터 순서대로 하나씩 중요한 단어인지 판단합니다.
무지가 당신에게 보내온 메시지를 나타내는 문자열 message와 스포 방지가 적용된 구간을 나타내는 2차원 정수 배열 spoiler_ranges가 매개변수로 주어질 때, 스포 방지 단어 중 중요한 단어의 수를 return 하도록 solution 함수를 완성해 주세요.

제한사항
1 ≤ message의 길이 ≤ 20,000
message는 알파벳 소문자, 숫자 그리고 공백으로 이루어져 있습니다.
message는 하나 이상의 단어로 구성된 문자열입니다.
공백은 연속해서 등장하지 않습니다.

0 ≤ start ≤ end < message의 길이
모든 구간은 서로 겹치지 않으며, start 기준으로 오름차순 정렬되어 주어집니다.

```python
def solution(message, spoiler_ranges):
    # ──────────────────────────────────────────────────────────────
    # [핵심 아이디어]
    # 1) 메시지를 단어로 나누고, 단어마다 (시작 인덱스, 끝 인덱스, 문자열)을 기록한다.
    # 2) 단어마다 "스포 방지 구간과 한 글자라도 겹치는지" 확인해서
    #    - 겹치면    → 스포 방지 단어
    #    - 안 겹치면 → 일반 단어 (= 스포 방지 구간이 아닌 곳에 등장한 단어)
    # 3) 스포 방지 단어를 왼쪽부터 보면서
    #    - 일반 단어 집합에 없고
    #    - 이전에 공개된 스포 방지 단어 집합에도 없으면 → 중요한 단어
    #
    # [왜 단어 순서대로만 보면 되나?]
    # 구간은 start 기준으로 정렬되어 있고 서로 겹치지 않는다.
    # 그래서 구간을 왼쪽부터 클릭하면 단어도 왼쪽부터 "완전히 공개"된다.
    # 여러 구간에 걸친 단어는 마지막 구간을 클릭할 때 공개되는데,
    # 그보다 왼쪽에 있는 단어들은 이미 그 전에 공개되어 있다.
    # 따라서 공개 순서 = 단어의 위치 순서이다.
    # ──────────────────────────────────────────────────────────────

    # ---------- 1. 단어 분리 (인덱스 정보 포함) ----------
    # split()을 쓰면 원래 인덱스를 잃기 때문에 직접 훑는다.
    words = []                      # (start, end, word)
    i, n = 0, len(message)
    while i < n:
        if message[i] == ' ':       # 공백은 건너뜀
            i += 1
            continue
        s = i                       # 단어 시작
        while i < n and message[i] != ' ':
            i += 1
        words.append((s, i - 1, message[s:i]))   # end는 포함 인덱스

    # ---------- 2. 단어마다 스포 방지 여부 판정 (투 포인터) ----------
    # 단어 [s, e]와 구간 [a, b]가 겹친다  ⇔  a <= e 이고 s <= b
    # 단어와 구간이 모두 왼쪽→오른쪽으로 정렬되어 있으므로
    # 구간 포인터 j는 뒤로 돌아갈 필요가 없다 → 전체 O(N + M)
    normal = set()                  # 스포 방지 구간 밖에 등장한 단어들
    spoiler = []                    # 스포 방지 단어 (왼쪽부터 순서대로)
    j, m = 0, len(spoiler_ranges)
    for s, e, w in words:
        # 이 단어보다 완전히 왼쪽에서 끝나는 구간(b < s)은 버린다.
        # (뒤에 나올 단어는 더 오른쪽에 있으니 이 구간과 다시 겹칠 일이 없다)
        while j < m and spoiler_ranges[j][1] < s:
            j += 1
        # 이제 spoiler_ranges[j]는 b >= s 를 만족한다.
        # 이 구간의 시작 a가 단어 끝 e 이하이면 겹친다.
        if j < m and spoiler_ranges[j][0] <= e:
            spoiler.append(w)
        else:
            normal.add(w)
        # 참고: 한 구간이 여러 단어를 덮을 수 있으므로
        # 겹쳤다고 해서 j를 바로 넘기면 안 된다.

    # ---------- 3. 중요한 단어 세기 ----------
    revealed = set()                # 이미 공개된 스포 방지 단어
    answer = 0
    for w in spoiler:
        # 조건 1: 스포 방지 단어 → spoiler 리스트에 있으니 이미 만족
        # 조건 2: 스포 방지 구간이 아닌 곳에 등장한 적 없음 → w not in normal
        # 조건 3: 이전에 공개된 스포 방지 단어와 중복 아님 → w not in revealed
        if w not in normal and w not in revealed:
            answer += 1
        # 중요한 단어인지와 상관없이 "공개된 스포 방지 단어"로 기록한다.
        revealed.add(w)

    return answer
```

```python
def solution(message, spoiler_ranges):
    # ---------- 1. 가려진 글자 표시하기 ----------
    # 글자마다 "가려졌나?"를 True/False로 기록한다.
    hidden = [False] * len(message)
    for start, end in spoiler_ranges:
        for k in range(start, end + 1):      # end도 포함
            hidden[k] = True

    # ---------- 2. 단어를 하나씩 보면서 분류하기 ----------
    normal = set()      # 한 글자도 안 가려진 단어
    spoiler = []        # 한 글자라도 가려진 단어 (왼쪽부터 순서대로)

    idx = 0                                   # 지금 단어의 시작 위치
    for w in message.split(' '):
        start = idx
        end = idx + len(w) - 1
        idx += len(w) + 1                     # 다음 단어 위치 (+1은 공백)

        if w == '':                           # 맨 앞/뒤 공백 대비
            continue

        if any(hidden[start:end + 1]):        # 한 글자라도 가려졌으면
            spoiler.append(w)
        else:
            normal.add(w)

    # ---------- 3. 중요한 단어 세기 ----------
    revealed = set()    # 이미 공개된 스포 단어
    answer = 0
    for w in spoiler:
        if w not in normal and w not in revealed:
            answer += 1
        revealed.add(w)

    return answer
```

```python
def solution(message, spoiler_ranges):
    answer = 0
    
    # 1. spoiler_ranges 를 각 문자마다 True False로 표시하기
    hidden = [False] * len(message)
    for s, e in spoiler_ranges:
        for i in range(s,e+1):
            hidden[i] = True
        
            
    # 2. 모든 단어에 대해서 스포 방지 단어와 일반 단어 구분하기
    sw = []
    nw = set()
    
    si = 0
    for w in message.split(" "):
        if w == " ": continue # 앞 뒤 공백 문자 건너뛰기
        
        ei = si + len(w) - 1
        
        # 하나라도 스포 구간이라면 스포 단어
        if any(hidden[si:ei+1]):
            sw.append(w)
        else:
            nw.add(w) # 스포 구간 아니라면 일반 단어
            
        
        si = ei + 2
        
    # 3. sw를 돌며 nw 아니면서 한 번 등장하지 않았다면 중요 단어
    rw = set()
    aw = set()
    for w in sw:
        if w not in nw and w not in rw:
            rw.add(w)
            aw.add(w)
            
            
    answer = len(aw)
    print(aw)
    return answer
```
