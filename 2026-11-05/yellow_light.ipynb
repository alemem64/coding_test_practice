# 배운 점
1. 이중 리스트 만들 때 lst = [[]] * n 하면 안 됨. 얕은 복사라서 lst[0] 바꾸면 lst[1]도 바뀜. [[] for _ in range(n)]로 해야 함.

```python
def solution(signals):
    answer = 0
    n = len(signals)
    
    # 신호 총합의 모든 곱까지가 최대 시간
    maxt = 1
    for signal in signals:
        sigsum = 0
        for sig in signal:
            sigsum +=sig
        maxt *= sigsum
    # print(maxt)
    
        
    # 최대 시간까지 각 신호등에서 노란불을 True로 표시
    ylst = [[] for _ in range(n)]
    
    # print(ylst)
    

    for i in range(n):
        
        gt, yt, rt = signals[i]
        t1 = 1
        
        while t1 <= maxt:
            
            ylst[i] += [0]*gt + [1]*yt + [0]*rt
            t1 += gt + yt + rt
        


    # 모든 신호등 True이면 answer
    t2 = 0

    while t2 < maxt:
        is_all_y = 1
        for i in range(n):
            is_all_y *= ylst[i][t2]
        
            
        if is_all_y == 1:
            answer = t2 + 1
            return answer
        
            
        t2 +=1
        
    answer = -1
        
    
    
    return answer
```

```python
from math import lcm

def solution(signals):
    # 1. 확인할 최대 시간 = 각 신호등 주기(G+Y+R)의 최소공배수
    #    이 시간이 지나면 모든 신호등이 동시에 처음 상태로 돌아가서
    #    같은 패턴이 반복된다. 그 안에 답이 없으면 영원히 없다.
    limit = lcm(*(sum(s) for s in signals))

    # 2. 1초부터 limit초까지 하나씩 확인
    for t in range(1, limit + 1):

        # 3. t초에 모든 신호등이 노란불인지 확인
        #    (t-1) % 주기 → 한 주기 안에서 몇 번째 초인지 (0부터 시작)
        #    0 ~ G-1       : 초록불
        #    G ~ G+Y-1     : 노란불  ← 이 구간이면 True
        #    G+Y ~ 주기-1  : 빨간불
        if all(g <= (t - 1) % (g + y + r) < g + y for g, y, r in signals):
            return t

    # 4. 끝까지 없으면 -1
    return -1
```

```python
def solution(signals):
    # 1. 언제까지 확인할지 정하기: 주기를 전부 곱한 값
    maxt = 1
    for g, y, r in signals:
        maxt *= (g + y + r)

    # 2. 1초부터 maxt초까지 하나씩 확인
    for t in range(1, maxt + 1):

        # 3. 지금(t초) 노란불인 신호등 개수 세기
        yellow_count = 0
        for g, y, r in signals:
            cycle = g + y + r          # 한 바퀴 길이
            pos = (t - 1) % cycle      # 한 바퀴 안에서 몇 번째 초인지 (0부터)
            if g <= pos < g + y:       # 초록 g초가 지나고, 노란 y초 동안
                yellow_count += 1

        # 4. 노란불 개수 == 신호등 개수 → 전부 노란불
        if yellow_count == len(signals):
            return t

    # 5. 끝까지 없으면 -1
    return -1
```
